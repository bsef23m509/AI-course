## Step 1

Complete the Knowledge Base class.

```python
class HospitalKB:
    def __init__(self):
        self.patients = []
        self.doctors = []
        self.treatments = []
        self.assignments = set()
        self.critical = set()

    def add_patient(self, p):
        self.patients.append(p)

    def add_doctor(self, d):
        self.doctors.append(d)

    def add_treatment(self, t):
        self.treatments.append(t)

    def add_assignment(self, p, t):
        self.assignments.add((p, t))

    def add_critical(self, t):
        self.critical.add(t)

    # Predicates
    def Patient(self, p):
        return p in self.patients

    def Doctor(self, d):
        return d in self.doctors

    def Treatment(self, t):
        return t in self.treatments

    def Assigned(self, p, t):
        return (p, t) in self.assignments

    def Critical(self, t):
        return t in self.critical
```

## Step 2

Implement file readers.

```python
def read_patients(file, kb):
    with open(file, 'r') as f:
        for line in f:
            if not line.strip() or line.startswith("#"):
                continue
            kb.add_patient(line.strip())


def read_doctors(file, kb):
    with open(file, 'r') as f:
        for line in f:
            if not line.strip() or line.startswith("#"):
                continue
            kb.add_doctor(line.strip())


def read_treatments(file, kb):
    with open(file, 'r') as f:
        for line in f:
            if not line.strip() or line.startswith("#"):
                continue
            kb.add_treatment(line.strip())


def read_assignments(file, kb):
    with open(file, 'r') as f:
        for line in f:
            if not line.strip() or line.startswith("#"):
                continue
            p, t = line.split()
            kb.add_assignment(p.strip(), t.strip())


def read_critical(file, kb):
    with open(file, 'r') as f:
        for line in f:
            if not line.strip() or line.startswith("#"):
                continue
            kb.add_critical(line.strip())
```

## Step 3

Implement First-Order Logic Evaluations.

```python
class FOLEvaluator:
    def __init__(self, kb):
        self.kb = kb

    def critical_patients_are_valid(self):
        """
        ∀p (Assigned(p, T1) → Patient(p))
        """
        for (p, t) in self.kb.assignments:
            if t == "T1" and not self.kb.Patient(p):
                return False
        return True

    def exists_patient_with_treatment(self):
        """
        ∃t (Treatment(t) ∧ Assigned(Ali, t))
        """
        for t in self.kb.treatments:
            if self.kb.Treatment(t) and self.kb.Assigned("Ali", t):
                return True
        return False

    def every_patient_has_treatment(self):
        """
        ∀p (Patient(p) → ∃t Assigned(p, t))
        """
        for p in self.kb.patients:
            has_treatment = False
            for t in self.kb.treatments:
                if self.kb.Assigned(p, t):
                    has_treatment = True
                    break
            if not has_treatment:
                return False
        return True

    def exists_patient_with_no_treatment(self):
        """
        ∃p ∀t (Treatment(t) → ¬Assigned(p, t))
        """
        for p in self.kb.patients:
            has_any_treatment = False
            for t in self.kb.treatments:
                if self.kb.Assigned(p, t):
                    has_any_treatment = True
                    break
            if not has_any_treatment:
                return True
        return False

    def critical_treatments_have_patients(self):
        """
        ∀t (Critical(t) → ∃p Assigned(p, t))
        """
        for t in self.kb.critical:
            has_patient = False
            for p in self.kb.patients:
                if self.kb.Assigned(p, t):
                    has_patient = True
                    break
            if not has_patient:
                return False
        return True
```

## Step 4

Complete evaluation function.

```python
def evaluate(kb):
    evaluator = FOLEvaluator(kb)

    results = []

    results.append("critical_patients_are_valid: " + str(evaluator.critical_patients_are_valid()))
    results.append("exists_patient_with_treatment: " + str(evaluator.exists_patient_with_treatment()))
    results.append("every_patient_has_treatment: " + str(evaluator.every_patient_has_treatment()))
    results.append("exists_patient_with_no_treatment: " + str(evaluator.exists_patient_with_no_treatment()))
    results.append("critical_treatments_have_patients: " + str(evaluator.critical_treatments_have_patients()))

    return results
```

## Step 5: Output Writer

Do NOT modify.

```python
def write_output(results, filename="output.txt"):
    output = "FOL Evaluation Results\n"
    output += "----------------------\n"

    for r in results:
        output += r + "\n"

    print(output)

    with open(filename, "w") as f:
        f.write(output)
```
