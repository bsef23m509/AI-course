## Step 1 — Define Data Structures

We represent propositions and rules using classes.


```python
class Proposition:

    def __init__(self, name, value):
        self.name = name
        self.value = value


class CourseRule:

    def __init__(self, name, rule):
        self.name = name
        self.rule = rule

    def evaluate(self, prop_dict):

        """
        Pseudocode

        1. Take the rule string
        2. Replace proposition names with their True/False values
        3. Evaluate the logical expression
        4. Return the result (True/False)
        """

        logical_dict = { "AND": "and", "OR": "or", "NOT": "not" }

        r = self.rule

        for name, value in prop_dict.items():
            r = r.replace(name, value)
        
        for name, value in logical_dict.items():
            r = r.replace(name, value)
        
        ans = eval(r)

        return ans
```

## Step 2 — Implement Basic Logical Operators

Implement the following propositional logic operators.

```python
def logical_and(a, b):

    """
    Return the logical AND of a and b
    """

    return a and b


def logical_or(a, b):

    """
    Return the logical OR of a and b
    """

    return a or b


def logical_not(a):

    """
    Return the logical negation of a
    """

    return not a
```

## Step 3 — Advanced Logical Operators

Implement the following operators.

Implication:

P → Q

Equivalent to:

¬P OR Q


Equivalence:

P ↔ Q

True only if both have the same truth value.

```python
def implication(p, q):

    """
    Implement logical implication
    """

    return logical_or(logical_not(p), q)


def equivalence(p, q):

    """
    Implement logical equivalence
    """

    return p == q
```

## Step 4 — Read Input Files

Implement functions that read data from text files.

```python
def read_student(filename):

    """
    Return a dictionary containing proposition name and value
    """

    props = {}

    with open(filename, 'r') as f:
        for line in f:
            if not line.strip():
                continue

            
            name, value = line.strip().split()
            props[name] = value
            

    return props


def read_courses(filename):

    """
    Return a list containing CourseRule objects
    """

    courses = []

    with open(filename, 'r') as f:
        for line in f:
            if not line.strip():
                continue

            
            name, rule = line.strip().split(":")
            cr = CourseRule(name, rule.strip())
            courses.append(cr)
            

    return courses
```

## Step 5 — Evaluate Eligibility

Determine which courses the student can take.

```python
def eligible_courses(prop_dict, courses):

    """
    Pseudocode:

    1. Evaluate each course rule
    2. Add its name to a list if True
    3. Return the list.
    """

    true_list = []

    for c in courses:
        if c.evaluate(prop_dict):
            true_list.append(c.name)

    return true_list
```

## Step 6 — Truth Table Generator

Generate all combinations of truth values for given variables
and evaluate a logical formula.

```python
import itertools

def truth_table(variables, formula):

    """
    variables: list like ["P","Q","R"]

    Pseudocode

    1. Generate all combinations of True/False
    2. For each combination:
        assign truth values to variables
        evaluate the formula
        print the result
    """

    logical_dict = { "AND": "and", "OR": "or", "NOT": "not" }

    for name, value in logical_dict.items():
            formula = formula.replace(name, value)

    print(" \t| ".join(variables), " \t\t| Result")
    

    for values in itertools.product([True, False], repeat=len(variables)):
        f = formula

        value_dict = dict(zip(variables, values))

        for name, value in value_dict.items():
            f = f.replace(name, str(value))

        result = eval(f)
    
        print(" \t| ".join(list(map(str, values))), " \t| ", result)


truth_table(["p", "q"], "p AND q")
    
```

## Step 7 — Detect Tautology or Contradiction (Challenge)

A **tautology** is a formula that is always True.

Example:
P OR NOT P

A **contradiction** is a formula that is always False.

Example:
P AND NOT P

Write a function that checks whether a formula is:

• a tautology  
• a contradiction  
• or neither

```python
def analyze_formula(variables, formula):

    """
    Pseudocode

    1. Generate truth table for the formula
    2. Collect all results
    3. If all results True → tautology
    4. If all results False → contradiction
    5. Otherwise → contingent formula
    """

    logical_dict = { "AND": "and", "OR": "or", "NOT": "not" }

    for name, value in logical_dict.items():
            formula = formula.replace(name, value)

    results = []
    

    for values in itertools.product([True, False], repeat=len(variables)):
        f = formula

        value_dict = dict(zip(variables, values))

        for name, value in value_dict.items():
            f = f.replace(name, str(value))

        result = eval(f)
        results.append(result)

    if all(results):
        return "tautology"
    elif all([not r for r in results]):
        return "contradiction"
    else:
        return "contingent formula"

analyze_formula(["p", "q"], "p AND q")
```

## Step 8 — Output and Main Program

The following functions are provided and **must NOT be modified**.

`write_output()` will display the results and also save them to a file.

`main()` runs the full program using the functions you implemented earlier.

```python
def write_output(eligible_list, filename="output.txt"):

    result = "Eligible Courses\n"
    result += "----------------\n"

    if len(eligible_list) == 0:
        result += "No courses available\n"
    else:
        for course in eligible_list:
            result += course + "\n"

    # print to console
    print(result)

    # write to file
    with open(filename, "w") as f:
        f.write(result)



def main():

    student_props = read_student("student.txt")

    courses = read_courses("courses.txt")

    eligible = eligible_courses(student_props, courses)

    write_output(eligible)


if __name__ == "__main__":
    main()
```
