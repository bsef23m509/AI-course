# Task 1: Move All Zeroes to the Right

Problem: Given a list of integers, move all 0s to the end while maintaining the relative order of non-zero elements.

Example:
Input: [0,1,0,3,12]
Output: [1,3,12,0,0]

Constraints:
- Do it in-place
- Do not use extra lists for non-zero filtering

```python
def move_zeroes(nums):
    last_non_zero = 0  

    for i in range(len(nums)):
        if nums[i] != 0:
            nums[last_non_zero] = nums[i]
            last_non_zero += 1

    for i in range(last_non_zero, len(nums)):
        nums[i] = 0

    return nums

nums = [0, 1, 0, 3, 12]
print(move_zeroes(nums))
```

# Task 2: Minimum Number Using heapq

Problem: Given a list of integers, return the minimum element using Python’s heapq module.

Example:
Input: [5,3,8,1,2]
Output: 1

Constraints:
- Use heapq
- Do not use min()
- Time complexity should be O(n)

```python
import heapq

def find_min(nums):
    heapq.heapify(nums)
    return nums[0]

nums = [5,3,8,1,2]
print(find_min(nums.copy()))
```

# Task 3: Count Unique Elements Using Dictionary

Problem: Given a list of integers, count how many unique elements are present using a dictionary.

Example:
Input: [1,2,2,3,4,4,5]
Output: 5

Constraints:
- Use dictionary
- Do not use set()

```python
def count_unique(nums):
    occurrence = {}
    for num in nums:
        occurrence[num] = True
    return len(occurrence)

nums = [1, 2, 2, 3, 4, 4, 5]
print(count_unique(nums))
```

# Task 4: Indexes of Top 5 Maximum Elements (Without Sorting)

Problem: Given a list of integers, return the indexes of the 5 largest elements without sorting the entire list.

Example:
Input: [10,4,3,50,23,90]
Output: [5,3,4,0,1]

Constraints:
- Do NOT use sorted()
- Do NOT modify original list
- Preferably O(n log k)

```python
def top5_indexes(nums):
    indexes = []

    max_heap = []
    for i, num in enumerate(nums):
        heapq.heappush(max_heap, (-num, i))

    for _ in range(5):
        if max_heap:
            val, idx = heapq.heappop(max_heap)
            indexes.append(idx)
    
    return indexes

nums = [10,4,3,50,23,90]
print(top5_indexes(nums))
```

# Task 5: First Non-Repeating Character

Problem: Given a string, return the first character that appears only once.

Example:
Input: 'aabbcddbe'
Output: 'c'

If no such character exists return None.

Constraints:
- Use dictionary for frequency counting
- Time complexity O(n)

```python
def first_non_repeating(s):
    freq = {}

    for ch in s:
        if ch not in freq:
            freq[ch] = 1
        else:
            freq[ch] += 1

    for ch in s:
        if freq[ch] == 1:
            return ch

    return None

print(first_non_repeating("aabbcddbe"))
```

# Section 3: Stack Implementation

```python
class Stack:
    '''Implement Stack (LIFO)'''

    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if self.is_empty():
            return None
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            return None
        return self.items[-1]

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)
```

# Section 4: Queue Implementation

```python
class Queue:
    '''Implement Queue (FIFO)'''

    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if self.is_empty():
            return None
        return self.items.pop(0)

    def front(self):
        if self.is_empty():
            return None
        return self.items[0]

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)
```

# Section 5: Graph Traversal (BFS & DFS)

```python
traffic_graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'D'],
    'D': ['B', 'C']
}

def bfs(graph, start):
    '''Implement BFS manually using Queue class'''
    visited = set()
    queue = Queue()
    queue.enqueue(start)
    visited.add(start)
    order = []

    while not queue.is_empty():
        node = queue.dequeue()
        order.append(node)

        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.enqueue(neighbor)

    return order

def dfs(graph, start, visited=None):
    '''Implement DFS recursively'''
    if visited is None:
        visited = set()

    visited.add(start)
    order = [start]

    for neighbor in graph.get(start, []):
        if neighbor not in visited:
            order.extend(dfs(graph, neighbor, visited))

    return order
```

# Section 6: Grid Traversal (2D BFS)

```python
grid = [
    [0, 0, 1],
    [0, 1, 0],
    [0, 0, 0]
]

def grid_bfs(grid, start):
    rows = len(grid)
    cols = len(grid[0]) if rows else 0

    visited = [[False for _ in range(cols)] for _ in range(rows)]
    order = []

    directions = [(-1,0), (1,0), (0,-1), (0,1)]

    queue = Queue()
    sr, sc = start
    queue.enqueue((sr, sc))
    visited[sr][sc] = True

    while not queue.is_empty():
        r, c = queue.dequeue()
        order.append((r, c))

        for dr, dc in directions:
            nr, nc = r + dr, c + dc
            if 0 <= nr < rows and 0 <= nc < cols:
                if not visited[nr][nc] and grid[nr][nc] == 0:
                    visited[nr][nc] = True
                    queue.enqueue((nr, nc))

    return order
```

# Section 7: Min Heap & Max Heap
