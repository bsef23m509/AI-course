Create the following functions:

```python
def calculate_total(marks):
    return sum(marks)


def calculate_average(marks):
    return sum(marks) / len(marks)


def calculate_grade(avg):
    grade = 'A'

    if avg >= 85:
        grade = 'A'
    elif avg >= 80:
        grade = 'A-'
    elif avg >= 75:
        grade = 'B+'
    elif avg >= 70:
        grade = 'B'
    elif avg >= 65:
        grade = 'B-'
    elif avg >= 60:
        grade = 'C+'
    elif avg >= 55:
        grade = 'C'
    else:
        grade = 'F'

    return grade


def is_prime(n):
    for i in range(2, n):
        if n % i == 0:
            return False

    return True


def fibonacci(n):
    if (n == 0 or n == 1):
        return n
    return fibonacci(n-1) + fibonacci(n-2)
    


def factorial(n):
    if n <= 0:
        return 1
    return n * factorial(n - 1)


print (calculate_total([67, 90, 64, 79, 87]))
print (calculate_average([67, 90, 64, 79, 87]))
print (calculate_grade(calculate_average([67, 90, 64, 79, 87])))
print (is_prime(5))
print (fibonacci(5))
print (factorial(5))
```

# Section 4: Algorithms (Manual Implementation Required)

Implement:
- Linear Search
- Binary Search
- Bubble Sort

Do NOT use built-in sort() or sorted().

Also explain time complexity in Markdown below this section.

```python
def linear_search(students, target_name):
    for s in students:
        if s.name == target_name:
            return s
    return None


def bubble_sort(students):
    n = len(students)
    
    for i in range(n):
        for j in range(0, n - i - 1):
            if students[j].roll_no > students[j + 1].roll_no:
                # swap
                students[j], students[j + 1] = students[j + 1], students[j]
    
    return students


def binary_search(sorted_students, target_roll):
    left = 0
    right = len(sorted_students) - 1

    while left <= right:
        mid = (left + right) // 2

        if sorted_students[mid].roll_no == target_roll:
            return sorted_students[mid]
        
        elif sorted_students[mid].roll_no < target_roll:
            left = mid + 1
        
        else:
            right = mid - 1

    return None


linear_search(students, "adil")
bubble_sort(students)
binary_search(students, 9)
```

### Time Complexity Analysis

Linear Search runs in O(n)

Binary Search runs in O(log n)

Bubble Sort runs in O(n^2)

# Section 5: Object-Oriented Programming

Create:
- Class Student
- Class GraduateStudent (inherits from Student)

Demonstrate:
- Encapsulation
- Inheritance
- Method overriding
- __str__ method

```python
class Student:

    def __init__(self, name, roll_no, marks, course):
        # Private attributes
        self.__name = name
        self.__roll_no = roll_no
        self.__marks = marks
        self.__course = course

    def calculate_total(self):
        return sum(self.__marks)

    def calculate_average(self):
        return self.calculate_total() / len(self.__marks)

    def calculate_grade(self):
        avg = self.calculate_average()

        if avg >= 85:
            return 'A'
        elif avg >= 75:
            return 'B'
        elif avg >= 65:
            return 'C'
        else:
            return 'F'

    def display_details(self):
        print(self)

    def __str__(self):
        return f"""
        Name: {self.__name}
        Roll No: {self.__roll_no}
        Marks: {self.__marks}
        Course: {self.__course}
    """


class GraduateStudent(Student):

    def __init__(self, name, roll_no, marks, course, specialization):
        super().__init__(name, roll_no, marks, course)
        self.__specialization = specialization

    def calculate_grade(self):
        avg = self.calculate_average()

        if avg >= 85:
            return 'A'
        elif avg >= 75:
            return 'B'
        elif avg >= 65:
            return 'C'
        else:
            return 'F'


student1 = Student("Adil", 101, [80, 75, 90, 85, 70], "CS")
grad1 = GraduateStudent("Ali", 201, [88, 92, 85, 90, 87], "SE", "AI")

print("----- Undergraduate -----")
student1.display_details()

print("----- Graduate -----")
grad1.display_details()
```

```python
import json

filename = "students.txt"



students = [
    Student("Ali", 101, [85, 78, 92, 88, 76], "CS"),
    Student("Adil", 102, [90, 85, 88, 92, 87], "SE")
]

# TODO: Save data to file using try-except
try:
    with open(filename, "w") as f:
        json.dump([s.__str__() for s in students], f, indent=4)
except Exception as e:
    print("Error writing to file:", e)
else:
    print(f"Data successfully saved to {filename}")
finally:
    print("Save operation finished.")


# TODO: Load data from file using try-except
try:
    with open(filename, "r") as f:
        data = json.load(f)
        loaded_students = data
except FileNotFoundError:
    print(f"File {filename} not found!")
except json.JSONDecodeError:
    print(f"Error decoding JSON from {filename}")
else:
    print(f"Data successfully loaded from {filename}")
finally:
    print("Load operation finished.\n")


print("Loaded Students:")
for s in loaded_students:
    print(s)


class InvalidMarksError(Exception):
    def __init__(self, message="Marks must be between 0 and 100"):
        super().__init__(message)
```

# Section 8: Functional Programming

Demonstrate:
- list comprehension
- map()
- filter()
- reduce()

Apply them to student data.

```python
from functools import reduce

# Sample student data
students = [
    {"name": "Ali", "roll_no": 101, "marks": [85, 78, 92, 88, 76]},
    {"name": "Adil", "roll_no": 102, "marks": [90, 85, 88, 92, 87]},
    {"name": "Ahmed", "roll_no": 103, "marks": [70, 65, 80, 75, 68]},
    {"name": "Ayesha", "roll_no": 104, "marks": [60, 55, 58, 62, 65]}
]

# TODO: List comprehension example
totals = [sum(s["marks"]) for s in students]
print(totals)

# TODO: map() example
averages = list(map(lambda s: sum(s["marks"])/len(s["marks"]), students))
print("Average marks (map):", averages)

# TODO: filter() example
top_students = list(filter(lambda s: sum(s["marks"])/len(s["marks"]) >= 80, students))
print("Top students (filter):", [s["name"] for s in top_students])

# TODO: reduce() example
total_all_marks = reduce(lambda acc, s: acc + sum(s["marks"]), students, 0)
print("Total marks of all students (reduce):", total_all_marks)
```
